# 01 - Exploratory Data Analysis (EDA)

In this notebook, we look at the dataset before building a machine-learning model.

```python
import pandas as pd
import matplotlib.pyplot as plt

data = pd.read_csv("../dataset/diabetes.csv")
data.head()
```

```python
print("Rows and columns:", data.shape)
data.info()
```

```python
# Check for missing values in every column.
data.isnull().sum()
```

```python
# Summary statistics for the number columns.
data.describe()
```

```python
# Count people with and without diabetes.
print(data["diabetes"].value_counts())

data["diabetes"].value_counts().plot(kind="bar", color=["skyblue", "salmon"])
plt.title("Diabetes class count")
plt.xlabel("Diabetes (0 = No, 1 = Yes)")
plt.ylabel("Number of people")
plt.show()
```

```python
# A histogram helps us see how ages are distributed.
data["age"].plot(kind="hist", bins=30, color="mediumpurple", edgecolor="black")
plt.title("Age distribution")
plt.xlabel("Age")
plt.show()
```
