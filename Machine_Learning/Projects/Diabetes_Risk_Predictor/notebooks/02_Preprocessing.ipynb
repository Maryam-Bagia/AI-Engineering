# 02 - Preprocessing

Machine-learning models need numbers. This notebook separates inputs from the target and turns text categories into numbers.

```python
import pandas as pd
from sklearn.model_selection import train_test_split

data = pd.read_csv("../dataset/diabetes.csv")
data.head()
```

```python
# X contains the information used for prediction. y contains the answer.
X = data.drop(columns="diabetes")
y = data["diabetes"]

print("X shape:", X.shape)
print("y shape:", y.shape)
```

```python
# Convert text categories such as Female/Male into 0 and 1 columns.
X_encoded = pd.get_dummies(X, columns=["gender", "smoking_history"], dtype=int)
X_encoded.head()
```

```python
# Keep 20% of the data for testing. stratify keeps the class ratio similar.
X_train, X_test, y_train, y_test = train_test_split(
    X_encoded, y, test_size=0.20, random_state=42, stratify=y
)

print("Training rows:", X_train.shape[0])
print("Test rows:", X_test.shape[0])
```
